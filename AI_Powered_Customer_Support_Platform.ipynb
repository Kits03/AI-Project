```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# 1. Load Data (Using a sample DataFrame for demonstration; replace with pd.read_csv('your_file.csv'))
data = {
    'ticket': [
        'Where is my package, it is delayed?',
        'I need a refund for my last order immediately.',
        'How do I reset my account password?',
        'My tracking number is not working.',
        'I was charged twice for the same item, please fix.',
        'Can you help me update my billing address?'
    ],
    'category': [
        'Shipping',
        'Billing',
        'Technical Support',
        'Shipping',
        'Billing',
        'Billing'
    ]
}
df = pd.DataFrame(data)

# 2. Prepare features (X) and labels (y)
X = df['ticket']
y = df['category']

# 3. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

# 4. Convert text into numerical vectors using TF-IDF
vectorizer = TfidfVectorizer()
X_train_vector = vectorizer.fit_transform(X_train)
X_test_vector = vectorizer.transform(X_test)

# 5. Train the Logistic Regression model
model = LogisticRegression()
model.fit(X_train_vector, y_train)

# 6. Evaluate accuracy on the test set
accuracy = model.score(X_test_vector, y_test)
print(f'Model Accuracy: {accuracy * 100:.2f}%\n')

# 7. Test with a custom new sentence
custom_ticket = ["My order is delayed and I want to check tracking"]
custom_vector = vectorizer.transform(custom_ticket)
predicted_category = model.predict(custom_vector)

print(f"Customer Ticket: '{custom_ticket[0]}'")
print(f"Predicted Category: --> {predicted_category[0]} <--")
```
